# TMDB EDA Overview

Use this notebook to generate presentation-ready visuals for the movie recommendation project.

## Recommended visuals
- Top genres bar chart
- Release trend area chart
- Rating histogram
- Runtime distribution
- Popularity vs. rating scatter

```python
from movie_recommendation import load_movie_data
import plotly.express as px

df = load_movie_data('..')
df.head()
```

```python
genre_counts = df['genres_list'].explode().value_counts().reset_index()
genre_counts.columns = ['genre', 'count']
px.bar(genre_counts.head(12), x='count', y='genre', orientation='h', title='Top Genres')
```

```python
yearly = df.dropna(subset=['release_year']).groupby('release_year').size().reset_index(name='count')
px.area(yearly, x='release_year', y='count', title='Movies Released Per Year')
```

```python
px.histogram(df, x='vote_average', nbins=20, title='Rating Distribution')
```
